Importing the Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn import metrics

```

Data Collection and Processing

```python
# loading the csv data to a pandas DataFrame
gold_data = pd.read_csv('/content/gld_price_data.csv')
```

```python
# print first 5 rows of the dataframe
gold_data.head()
```

```python
# print last 5 rows of the dataframe
gold_data.tail()
```

```python
# number of rows and columns
gold_data.shape
```

```python
# getting some basic information about the data
gold_data.info()
```

```python
# checking for missing values
gold_data.isnull().sum()
```

```python
# getting the ststistical measures of the data

gold_data.describe()
```

```python
gold_data_numeric = gold_data.drop('Date', axis=1)
correlation = gold_data_numeric.corr()
```

```python
# constructing a heatmap to understand the correlation
plt.figure(figsize=(8,8))
sns.heatmap(correlation,cbar = True,square=True,fmt = '.1f',annot = True,annot_kws={'size':8},cmap = 'Greens')
```

```python
#correlation values of GLD
print(correlation['GLD'])
```

```python
# checking the distribution of gold price
sns.histplot(gold_data_numeric['GLD'],color='red')
```

Splitting the Features and Target


```python
X = gold_data.drop(['GLD','Date'],axis=1)
Y = gold_data['GLD']
```

```python
print(X)
```

```python
print(Y)
```

Splitting into Training Data and Test Data


```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=2)
```

Model Training


```python
regressor = RandomForestRegressor(n_estimators=100)
```

```python
# training the model
regressor.fit(X_train,Y_train)
```

Model Evaluation


```python
test_data_pred = regressor.predict(X_test)
```

```python
print(test_data_pred)
```

```python
# R Squared Error
error_score = metrics.r2_score(Y_test,test_data_pred)
```

```python
print("R Squared Error : ",error_score)
```

Compare Actual Values and Predicted Values in a plot


```python
Y_test = list(Y_test)

```

```python
plt.plot(Y_test,color = 'blue',label = 'Actual values')
plt.plot(test_data_pred,color = 'red',label = 'Predicted values')
plt.title('Actual Price vs Predicted Price')
plt.xlabel('Number of Values')
plt.ylabel('GLD Price')
plt.legend()
plt.show()
```

```python
# Create a tuple of the current market values
# Order must be: SPX, USO, SLV, EUR/USD
input_data = (7137.90, 129.40, 70.37, 1.1706)

# Change input_data to a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# Reshape the array as we are predicting for one instance
input_data_reshaped = input_data_as_numpy_array.reshape(1,-1)

# Make the prediction
prediction = regressor.predict(input_data_reshaped)

print(f"Current Actual Gold Price (GLD): $434.38")
print(f"Model's Predicted Price: ${prediction[0]:.2f}")

# Calculate the difference
difference = abs(434.38 - prediction[0])
print(f"Difference: ${difference:.2f}")
```
